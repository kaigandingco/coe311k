```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
A = np.array([[-3.0, 2.0],
              [-1.0, 0.0]]) # Constructs the coefficient matrix for Euler's Method
y_0 = np.array([2.0, 1.0])

step_sizes = np.linspace(0.01, 0.2, 5)
num_steps = int(50)

plt.figure()
```

```python
for h in step_sizes:

    # Euler amplification matrix
    E = np.eye(A.shape[0]) + h * A

    eigenvalues = np.linalg.eigvals(E)
    radius = np.max(np.abs(eigenvalues))

    print(f"Step Size, h: {h:.3f}")
    print(f"Eigenvalues of Euler matrix: {eigenvalues}")

    if radius < 1:
        print("Solution is Stable\n")
    else:
        print("Solution may not be Entirely Stable ;)\n")
        
    Y = np.zeros((num_steps, len(y_0)))
    Y[0] = y_0

    for n in range(num_steps - 1):
        Y[n + 1] = E @ Y[n]

    X = np.arange(num_steps)

    plt.plot(X, Y[:, 0], label=f"y1, h={h:.2f}")
    plt.plot(X, Y[:, 1], label=f"y2, h={h:.2f}")

plt.title("Euler's Method Stability Analysis")
plt.xlabel("Step Number")
plt.ylabel("Value of y")
plt.legend()
plt.grid()
plt.show()
```
