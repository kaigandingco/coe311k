```python
import numpy as np
import time

def jacobi(A, b, x0, tolerance, maxIterations):
    n = len(b)
    x = x0.copy()
    
    for k in range (maxIterations):
        xNew = np.zeros_like(x)
        for i in range(n): 
            s = sum(A[i][j] * x[j] for j in range(n) if j!= i)
            xNew[i] = (b[i] - s) / A[i][i]
            
        if np.linalg.norm(xNew - x, ord=np.inf) < tolerance:
            return xNew, k
        
        x = xNew
    return x, maxIterations
```

```python
def gauss_seidel(A, b, x0, tolerance, maxIterations):
    n = len(b)
    x = x0.copy()
    
    for k in range (maxIterations):
        xNew = x.copy()
        for i in range(n):
            s1 = sum(A[i][j] * xNew[j] for j in range (i))
            s2 = sum(A[i][j] * x[j] for j in range(i + 1, n))
            xNew[i] = (b[i] - s1 -s2) / A[i][i]
            
        if np.linalg.norm(xNew - x, ord=np.inf) < tolerance:
            return xNew, k
        
        x = xNew
    return x, maxIterations
```

```python
# Challenge 6a, Jacobi

A = np.array([[10, -1, 2, 0, 0],
              [-1, 11, -1, 3, 0],
              [2, -1, 10, -1, 0],
              [0, 3, -1, 8, -2],
              [0, 0, 0, -2, 9]])
b = np.array([14, 30, 26, 25, 37])
x0 = np.zeros_like(b)
tolerance = 1e-6
maxIterations = 100

startTime = time.perf_counter()
solution, iterations = jacobi(A, b, x0, tolerance, maxIterations)
endTime = time.perf_counter()
totalTime = endTime - startTime

print("Challenge 6a, Jacobi\n")
print(f"Solution: {solution}")
print(f"Iterations: {iterations}")
print(f"Time: {duration:.6f} seconds")
```

```python
# Challenge 6a, Gauss-Seidel

A = np.array([[10, -1, 2, 0, 0],
              [-1, 11, -1, 3, 0],
              [2, -1, 10, -1, 0],
              [0, 3, -1, 8, -2],
              [0, 0, 0, -2, 9]])
b = np.array([14, 30, 26, 25, 37])
x0 = np.zeros_like(b)
tolerance = 1e-6
maxIterations = 100

startTime = time.perf_counter()
solution, iterations = gauss_seidel(A, b, x0, tolerance, maxIterations)
endTime = time.perf_counter()
duration = endTime - startTime

print("Challenge 6a, Gauss-Seidel\n")
print(f"Solution: {solution}")
print(f"Iterations: {iterations}")
print(f"Time: {duration:.6f} seconds")
```
