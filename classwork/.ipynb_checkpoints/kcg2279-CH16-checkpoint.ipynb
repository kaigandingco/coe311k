```python
import numpy as np
import matplotlib.pyplot as plt

# I hope the way I formatted this isn't too difficult to follow! I modified
# the demo code a little bit because I wanted to format it my way :/
```

```python
step_sizes = [0.01, 0.02, 0.03, 0.05, 0.1, 0.2, 0.3, 0.5]
plt.figure(figsize=(10, 5))

for h in step_sizes:
    
    # Defining parameters
    k = 1.0        # Constant in the equation (e.g., spring constant)
    x_start = 0    # Start of the interval
    x_end = 10     # End of the interval
    y_0 = 1.0      # Initial condition for y
    v_0 = 0.0      # Initial condition for dy/dx
    # Storing values
    X = np.arange(x_start, x_end + h, h)
    Y = np.zeros(len(X))
    Y[0] = y_0
    V = np.zeros(len(X))
    V[0] = v_0
    
    # Euler's Method
    for i in range(0, len(X)-1):
        xNew = X[i]
        yNew = Y[i]
        vNew = V[i]
        Y[i+1] = yNew + h * vNew
        V[i+1] = vNew + h * (-k * yNew)
        
    # Plotting
    plt.plot(X, Y, label=f'Position, y, with a Step Size, h, of {h}')
    plt.plot(X, V, label=f'Velocity, v, with a Step Size, h, of {h}', linestyle='--')

plt.title("Solutions of d²y/dx² = -ky using Euler's Method")
plt.xlabel('x')
plt.ylabel("Position (y) and Velocity (v)")
plt.legend(bbox_to_anchor = (0.5, -0.2))
plt.grid()
plt.show()
```

It becomes unstable beyond a step size of 0.3. After this point, the position and velocity curves show the most dramatic difference compared with the other curves in the plot. The curves for a step size of 0.5 have much more extreme oscillations. This suggests that the solutions become increasingly unstable as step size increases.

```python
k_values = np.arange(1, 11)
plt.figure(figsize=(10, 5))

for k in k_values:
    h = 0.1        # Step size
    x_start = 0    # Start of the interval
    x_end = 10     # End of the interval
    y_0 = 1.0      # Initial condition for y
    v_0 = 0.0      # Initial condition for dy/dx
    # Storing values
    X = np.arange(x_start, x_end + h, h)
    Y = np.zeros(len(X))
    Y[0] = y_0
    V = np.zeros(len(X))
    V[0] = v_0
    
    # Euler's Method
    for i in range(0, len(X)-1):
        xNew = X[i]
        yNew = Y[i]
        vNew = V[i]
        Y[i+1] = yNew + h * vNew
        V[i+1] = vNew + h * (-k * yNew)
        
    # Plotting
    plt.plot(X, Y, label=f'Position (y) when h={h} and k={k}')
    plt.plot(X, V, label=f'Velocity (v) when h={h} and k={k}', linestyle='--')
 
plt.title("Solutions of d²y/dx² = -ky using Euler's Method")
plt.xlabel('x')
plt.ylabel("Position (y) and Velocity (v)")
plt.legend(bbox_to_anchor = (0.5, -0.2))
plt.grid()
plt.show()
```

As the value of k increases, the amplitude of the wave increases, resulting in more extreme oscillations.

```python
k = 10
h = 0.01
x_start = 0    # Start of the interval
x_end = 10     # End of the interval
y_0 = 1.0      # Initial condition for y
v_0 = 0.0      # Initial condition for dy/dx
# Storing values
X = np.arange(x_start, x_end + h, h)
Y = np.zeros(len(X))
Y[0] = y_0
V = np.zeros(len(X))
V[0] = v_0

# Euler's Method
for i in range(0, len(X)-1):
        xNew = X[i]
        yNew = Y[i]
        vNew = V[i]
        Y[i+1] = yNew + h * vNew
        V[i+1] = vNew + h * (-k * yNew)

# Plot the results
plt.figure(figsize=(10, 5))
plt.plot(X, Y, label=f'Position (y) when h={h} and k={k}', color='hotpink')
plt.plot(X, V, label=f'Velocity (v) when h={h} and k={k}', color='limegreen', linestyle='--')
plt.title("Solutions of d²y/dx² = -ky using Euler's Method")
plt.xlabel('x')
plt.ylabel("Position (y) and Velocity (v)")
plt.legend(bbox_to_anchor = (0.5, -0.2))
plt.grid()
plt.show()
```

Euler's Method with a larger h value would not be effective in solving this because the stability of the representation is inversely proportional to the step size. Using Euler's Method, a larger step size results in a more unstable and less accurate result.
