```python
import random
import numpy as np
import matplotlib.pyplot as plt

# Calculating pi with the Monte Carlo method
def monte_carlo(numSamples):
    x = np.random.uniform(-1,1,size=numSamples)
    y = np.random.uniform(-1,1,size=numSamples)
    counter = ((x*x + y*y) < 1)
    return 4 * (np.sum(counter)/numSamples)
```

```python
absoluteError = []
relativeError = []

for z in range (0, 100):
    computedPi = monte_carlo(1000)
    
    # Absolute error
    absoluteError.append(np.abs(np.pi - computedPi)) 
    
    # Relative error
    relativeError.append(np.abs((np.pi - computedPi)/np.pi))
    
# Plotting absolute error
plt.subplot(1,2,1)
plt.plot(absoluteError, color="hotpink")
plt.title("Absolute Error")

# Plotting relative error
plt.subplot(1,2,2)
plt.plot(relativeError, color="dodgerblue")
plt.title("Relative Error")
plt.show()
```
