```python
import numpy as np

def jacobi(A, b, x0, tolerance, maxIterations):
    n = len(b)
    x = x0.copy()
    
    for k in range (maxIterations):
        xNew = np.zeros_like(x)
        for i in range(n): 
            s = sum(A[i][j] * x[j] for j in range(n) if j!= i)
            xNew[i] = (b[i] - s) / A[i][i]
            
        if np.linalg.norm(xNew - x, ord=np.inf) < tolerance:
            return xNew, k
        
        x = xNew
    return x, maxIterations
```

```python
def gauss_seidel(A, b, x0, tolerance, maxIterations):
    n = len(b)
    x = x0.copy()
    
    for k in range (maxIterations):
        xNew = x.copy()
        for i in range(n):
            s1 = sum(A[i][j] * xNew[j] for j in range (i))
            s2 = sum(A[i][j] * x[j] for j in range(i + 1, n))
            xNew[i] = (b[i] - s1 -s2) / A[i][i]
            
        if np.linalg.norm(xNew - x, ord=np.inf) < tolerance:
            return xNew, k
        
        x = xNew
    return x, maxIterations
```

```python
# Example 1, Jacobi

A = np.array([
             [0.582745,     0.48,    0.10,       0.,       0.],
             [    0.48, 1.044129,    0.46,     0.10,       0.],
             [    0.10,     0.46, 1.10431,     0.44,     0.10],
             [      0.,     0.10,    0.44, 0.963889,     0.42],
             [      0.,       0.,    0.10,     0.42, 0.522565]], dtype=float)
b = np.array([1.162745, 2.084129, 2.20431, 1.923889, 1.042565], dtype=float)
x0 = np.zeros_like(b)
tolerance = 1e-6
maxIterations = 100
solution, iterations = jacobi(A, b, x0, tolerance, maxIterations)

print("Example 1, Jacobi\n")
print(f"Solution: {solution}")
print(f"Iterations: {iterations}")
```

```python
# Example 1, Gauss-Seidel

A = np.array([
             [0.582745,     0.48,    0.10,       0.,       0.],
             [    0.48, 1.044129,    0.46,     0.10,       0.],
             [    0.10,     0.46, 1.10431,     0.44,     0.10],
             [      0.,     0.10,    0.44, 0.963889,     0.42],
             [      0.,       0.,    0.10,     0.42, 0.522565]], dtype=float)
b = np.array([1.162745, 2.084129, 2.20431, 1.923889, 1.042565], dtype=float)
x0 = np.zeros_like(b)
tolerance = 1e-6
maxIterations = 100
solution, iterations = gauss_seidel(A, b, x0, tolerance, maxIterations)

print("Example 1, Gauss-Seidel\n")
print(f"Solution: {solution}")
print(f"Iterations: {iterations}")
```
