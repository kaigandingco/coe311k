```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Part 1: Given Points
X = np.array([  1,   2,   3,   4,   5])
Y = np.array([1.2, 2.8, 3.6, 4.5, 5.1])
A = np.vstack([X, np.ones(len(X))]).T
m, c = np.linalg.lstsq(A, Y, rcond=None)[0]

print(f"Slope: {m}, Intercept: {c}")
plt.figure()
plt.scatter(X, Y)
plt.plot(X, m * X + c)
plt.title("Least Squares Method with Given Points")
plt.grid()
plt.show()
```

```python
# Part 2: Randomized Vectors
X = np.random.randint(0, 10, size=5)
Y = np.random.randint(0, 10, size=5)
A = np.vstack([X, np.ones(len(X))]).T
m, c = np.linalg.lstsq(A, Y, rcond=None)[0]

print(f"Slope: {m}, Intercept: {c}")
plt.figure()
plt.scatter(X, Y)
plt.plot(X, m * X + c)
plt.title("Least Squares Method with Random Points")
plt.grid()
plt.show()

"""
With random vectors, the best-fit line noticeably does not fit the
data points, especially in comparison with the best-fit line for the
given points in Part 1.
"""
```

```python
# Part 3: New Linear Equation (y = 5.3x + 2.4)
X = np.arange(0, 1, 0.01)
randPerturb = np.random.normal(0, 1, size=len(X))
Y = 5.3 * X + 2.4 + randPerturb
A = np.vstack([X, np.ones(len(X))]).T
m, c = np.linalg.lstsq(A, Y, rcond=None)[0]

print(f"Slope: {m}, Intercept: {c}")
plt.figure()
plt.scatter(X, Y)
plt.plot(X, m * X + c)
plt.title("Least Squares Method with Random Perturbation")
plt.grid()
plt.show()

"""
With the Least Squares method, the best-fit line is consistently
quite close to the original linear equation, despite the random
perturbation.
"""
```
