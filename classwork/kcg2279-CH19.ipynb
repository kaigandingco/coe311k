```python
import numpy as np
import matplotlib.pyplot as plt

def func(x, y):
    return -1000*y + 3000 - 2000*np.exp(-x)
```

```python
# Initial values and condition
all_h = [0.1, 0.01, 0.001] # RIP 0.0001 and 0.0000001
x_start = 0
x_end = 30
plt.figure()

for h in all_h:
    steps = int((x_end - x_start) / h)
    X = np.linspace(x_start, x_end, steps + 1)
    Y = np.zeros(steps + 1)
    Y[0] = 0
    
    for i in range(steps):
        Y[i + 1] = Y[i] + h * func(X[i], Y[i])
        
    # Plotting
    plt.plot(X, Y, label=f"Euler's Method Approximation with a Step Size of {h}")

# Exact solution
x_exact = np.linspace(0, 30, 10000)
exact = 3-0.998*np.exp(-1000*x_exact)-2.002*np.exp(-x_exact)

# Plotting
plt.plot(x_exact, exact, label="Exact Solution", color='hotpink')
plt.title("Euler's Forward Compared With the Analytical Solution")
plt.xlabel('x')
plt.ylabel('y')
plt.ylim(0, 30)
plt.legend()
plt.show()
```
