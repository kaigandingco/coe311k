## Section 1: Review of Part 2
In Part 2, I selected Option E, which models a population undergoing rapid birth-death turnover amidst slow environmental fluctuations, thus impacting carrying capacity. This results in a stiff ordinary differential equation. The governing ordinary differential equation is <br><br>$$\frac{dN}{dt} = -\lambda\ (N - N_{carrying}(t))$$<br>where $N(t)$ represents the population size at time $t$, $\lambda$ is the rate at which the population approaches equilibrium, and $N_{carrying}(t)$ is the carrying capacity, dependent on time and available resources. Using the given parameters for stiffness, $\lambda = 10^3$ indicates extremely rapid population turnover while $N_{carrying}(t)=10^6\ (1 + 0.5\ sin(0.1t))$ reflects a slowly varying environmental influence. The initial condition $N(0) = 10^5$ then indicates that the population begins below equilibrium. Together, these conditions reflect a rapidly fluctuating population experiencing periodic changes in resources.

The approximate stiffness ratio of this model is 10,000, meaning this system is very stiff. This high of a stiffness ratio suggests that the system's dynamics rely on vastly different time scales, with one part of the system changing around 10,000 times faster than another.

## Section 2: Newton-Raphson Damping Implementation
Damping and line search improve upon Newton-Raphson solving by controlling how far each step goes and only using step sizes that improve the solution model. Damping moves the solution in the same direction as the traditional Newton-Raphson step, but not as far. Line search is what determines how big that step should actually be. First, it takes the full step, and, if the full step does not improve the model, it then shrinks the step before trying again.

```python
# Implementation of damped NR solver

import numpy as np
import matplotlib.pyplot as plt

# Undamped Newton-Raphson solver
def basic_nr_solver(y_n, t_next, h, g_function, g_prime_function,
                    tolerance=1e-8, max_iterations=20, y_guess=None):
    
    if y_guess is None:
        y_guess = y_n

    residual_history = [] # Track convergence behavior

    for k in range(max_iterations):
        g = g_function(y_guess, y_n, t_next, h)
        residual_history.append(abs(g)) # Store magnitude of residual

        # Check convergence
        if abs(g) < tolerance:
            return y_guess, k + 1, True, residual_history

        g_prime = g_prime_function(y_guess, y_n, t_next, h)

        if abs(g_prime) < 1e-14:
            return y_guess, k + 1, False, residual_history

        delta = -g / g_prime # Direction

        # Full Newton step
        y_guess = y_guess + delta

    return y_guess, max_iterations, False, residual_history
```

```python
# Damped Newton-Raphson solver

def damped_nr_solver(y_n, t_next, h, g_function, g_prime_function,
                          tolerance=1e-8, max_iterations=20, alpha_min=1e-4,
                          max_backtracks=10, y_guess=None):

    if y_guess is None:
        y_guess = y_n

    residual_history = [] # Tracking convergence behavior
    alpha_history = [] # Stores damping factors
    backtrack_history = [] # Stores number of backtracking steps
    damping_needed = False # Flag if damping was used

    for k in range(max_iterations):
        g = g_function(y_guess, y_n, t_next, h)
        residual_history.append(abs(g))

        # Check convergence
        if abs(g) < tolerance:
            return {
                "solution": y_guess,
                "iterations": k + 1,
                "converged": True,
                "residual_history": residual_history,
                "alpha_history": alpha_history,
                "backtrack_history": backtrack_history,
                "damping_needed": damping_needed
            }

        g_prime = g_prime_function(y_guess, y_n, t_next, h)

        if abs(g_prime) < 1e-14:
            return {
                "solution": y_guess,
                "iterations": k + 1,
                "converged": False,
                "residual_history": residual_history,
                "alpha_history": alpha_history,
                "backtrack_history": backtrack_history,
                "damping_needed": damping_needed
            }

        delta = -g / g_prime # Direction

        alpha = 1.0
        g_current = abs(g)
        backtracks = 0

        # Line search
        for bt in range(max_backtracks):
            y_new = y_guess + alpha * delta
            g_new = abs(g_function(y_new, y_n, t_next, h))

            # Accepts the step if residual decreases
            if g_new < g_current:
                break

            # Reduces step size
            alpha = alpha / 2.0
            backtracks += 1
            damping_needed = True

            if alpha < alpha_min:
                return {
                    "solution": y_guess,
                    "iterations": k + 1,
                    "converged": False,
                    "residual_history": residual_history,
                    "alpha_history": alpha_history,
                    "backtrack_history": backtrack_history,
                    "damping_needed": damping_needed
                }

        y_guess = y_new

        alpha_history.append(alpha)
        backtrack_history.append(backtracks)

    return {
        "solution": y_guess,
        "iterations": max_iterations,
        "converged": False,
        "residual_history": residual_history,
        "alpha_history": alpha_history,
        "backtrack_history": backtrack_history,
        "damping_needed": damping_needed
    }
```

```python
# Demonstration (test problem) where damping helps

# Nonlinear residual
def g_nonlinear(y_next, y_current, t_next, h):
    return y_next**3 - 1

# Derivative of residual
def g_prime_nonlinear(y_next, y_current, t_next, h):
    return 3 * y_next**2

y_current = 0.0
t_next = 1.0
h_test = 10.0

# Poor initial guess intentionally chosen
initial_guess = 0.1

# Runs basic Newton-Raphson solver
basic_sol, basic_iters, basic_converged, basic_residuals = basic_nr_solver(
    y_current,
    t_next,
    h_test,
    g_nonlinear,
    g_prime_nonlinear,
    tolerance=1e-8,
    max_iterations=20,
    y_guess=initial_guess
)

# Runs damped Newton-Raphson solver
damped_result = damped_nr_solver(
    y_current,
    t_next,
    h_test,
    g_nonlinear,
    g_prime_nonlinear,
    tolerance=1e-8,
    max_iterations=20,
    alpha_min=1e-4,
    max_backtracks=10,
    y_guess=initial_guess
)

# Convergence comparison plot
plt.figure()
plt.plot(basic_residuals, color='hotpink', label="Basic Newton-Raphson Solving")
plt.plot(damped_result["residual_history"], color='limegreen', label="Damped Newton-Raphson Solving")
plt.title("Comparing Basic and Damped Newton-Raphson Convergence")
plt.xlabel("Iteration Number")
plt.ylabel("Residual, g(y)")
plt.legend()
plt.grid()
plt.show()

print("Basic (Undamped) Newton-Raphson")
print(f"Converged: {basic_converged}")
print(f"Iterations: {basic_iters}")
print(f"Approximate Solution: {basic_sol}")
print()

print("Damped Newton-Raphson")
print(f"Converged: {damped_result['converged']}")
print(f"Iterations: {damped_result['iterations']}")
print(f"Approximate Solution: {damped_result['solution']}")
print(f"Damping Needed: {damped_result['damping_needed']}")
print(f"Final Alpha Used: {damped_result['alpha_history'][-1] if damped_result['alpha_history'] else None}")
print(f"Backtracking Steps per Iteration: {damped_result['backtrack_history']}")
```

The convergence plot above shows that the basic, undamped Newton-Raphson method experiences a large increase in residual during early iterations due to an unnecessarily large step size and poor initial guess. In contrast, the damped Newton-Raphson method maintains a steadily decreasing residual by reducing the step size when necessary. This demonstrates that damping prevents overshooting and improves robustness, particularly when the initial guess is poor.

## Section 3: Adaptive Step Size Implementation
Step-doubling estimates error without knowing the exact solution by comparing two computations over the same interval. One computation uses a single, full-sized step while the other uses two smaller, half-sized steps. Because smaller step sizes generally produce more accurate results, the solution obtained from the two half-steps is treated as a better approximation, and the difference between these two results provides an estimate of the error in the larger step. 

```python
# Adaptive Step Size Control for Part 2 Population Dynamics

# Parameter definitions
lamb = 10**3 # rate at which population approaches equilibrium
N0 = 10**5 # initial condition
t0, t_final = 0, 10

# Carrying capacity function, dependent on time
def N_carrying(t):
    return 10**6 * (1 + 0.5 * np.sin(0.1 * t)) # slow time scale will be 1/0.1

# ODE definition
def f(t, N):
    return -lamb * (N - N_carrying(t))

# One backward Euler step
# Uses previous value as initial guess
def implicit_euler_step_NR(y_n, t_next, h, tolerance=1e-10, max_iterations=20):
    
    y_guess = y_n

    for k in range(1, max_iterations + 1):
        g = y_guess - y_n - h * f(t_next, y_guess)

        if abs(g) < tolerance:
            return y_guess, k

        g_prime = 1 + h * lamb

        y_guess = y_guess - g / g_prime

    return y_guess, max_iterations
```

```python
# Estimate local error using step-doubling.
# Take one full step of size h.
# Take two half-steps of size h/2.
# Compare the two results.
# The two-half-step solution is treated as more accurate.
    
def estimate_error_step_doubling(y_n, t_n, h, nr_solver):
    
    # One full step
    y_full, iters_full = nr_solver(y_n, t_n + h, h)

    # First half-step
    y_half_1, iters_1 = nr_solver(y_n, t_n + h / 2, h / 2)

    # Second half-step
    y_half, iters_2 = nr_solver(y_half_1, t_n + h, h / 2)

    # Relative error estimate
    scale = max(abs(y_half), 1.0)
    error = abs(y_half - y_full) / scale

    # Count all Newton-Raphson iterations used
    total_iters = iters_full + iters_1 + iters_2

    return y_full, y_half, error, total_iters
```

```python
# If error is large, reduce h.
# If error is small, increase h.
    
def adjust_step_size(h_current, error, tol, safety=0.9,
                     h_min=1e-6, h_max=1.0):

    if error < 1e-14:
        h_new = 2.0 * h_current
    else:
        h_new = h_current * np.sqrt(tol / error)

    # Applies safety factor
    h_new = safety * h_new

    # Enforces minimum and maximum step sizes
    h_new = max(h_min, min(h_max, h_new))

    # Prevents sudden extreme changes in h
    h_new = max(0.2 * h_current, min(5.0 * h_current, h_new))

    return h_new
```

```python
# Attempts one adaptive step
def adaptive_step(y_n, t_n, h, tol, nr_solver,
                  h_min=1e-6, h_max=1.0, safety=0.9):

    y_full, y_half, error, iters = estimate_error_step_doubling(
        y_n, t_n, h, nr_solver
    )

    h_new = adjust_step_size(
        h_current=h,
        error=error,
        tol=tol,
        safety=safety,
        h_min=h_min,
        h_max=h_max
    )

    if error < tol:
        # Accepts step and use the more accurate half-step solution
        return y_half, h, h_new, True, error, iters
    else:
        # Rejects step and try again with smaller h
        return y_n, h, h_new, False, error, iters
```

```python
# Adaptive Euler method using step-doubling
def adaptive_implicit_euler(y0, t0, t_final, h0, tol,
                            h_min=1e-6, h_max=1.0, safety=0.9):

    t_values = [t0]
    y_values = [y0]
    h_history = []
    error_history = []
    
    accepted_t = []
    rejected_t = []

    t = t0
    y = y0
    h = h0

    accepted_steps = 0
    rejected_steps = 0
    total_nr_iterations = 0

    while t < t_final:
        # Do not step past final time
        if t + h > t_final:
            h = t_final - t

        y_next, h_used, h_new, accepted, error, iters = adaptive_step(
            y_n=y,
            t_n=t,
            h=h,
            tol=tol,
            nr_solver=implicit_euler_step_NR,
            h_min=h_min,
            h_max=h_max,
            safety=safety
        )

        total_nr_iterations += iters

        if accepted:
            t = t + h_used
            y = y_next

            t_values.append(t)
            y_values.append(y)
            h_history.append(h_used)
            error_history.append(error)

            accepted_steps += 1
            accepted_t.append(t)

        else:
            rejected_steps += 1
            rejected_t.append(t)

        h = h_new

        if h <= h_min and not accepted:
            print("Warning: minimum step size reached and step still rejected.\n")
            break

    stats = {
        "accepted_steps": accepted_steps,
        "rejected_steps": rejected_steps,
        "total_steps_attempted": accepted_steps + rejected_steps,
        "total_nr_iterations": total_nr_iterations,
        "acceptance_rate": accepted_steps / (accepted_steps + rejected_steps),
        "accepted_t": accepted_t,
        "rejected_t": rejected_t
    }

    return (
        np.array(t_values),
        np.array(y_values),
        np.array(h_history),
        np.array(error_history),
        stats
    )
```

```python
# Fixed-step Euler's Backward Method for Comparison

def euler_backward_fixed(y0, t0, t_final, h):

    t_values = np.arange(t0, t_final + h, h)

    if t_values[-1] > t_final:
        t_values[-1] = t_final

    y_values = np.zeros(len(t_values))
    y_values[0] = y0

    total_iters = 0

    for i in range(len(t_values) - 1):
        h_step = t_values[i + 1] - t_values[i]
        y_values[i + 1], iters = implicit_euler_step_NR(
            y_values[i],
            t_values[i + 1],
            h_step
        )
        total_iters += iters

    return t_values, y_values, total_iters
```

```python
# Comparing fixed and adaptive methods

fixed_h = 0.01
tol = 1e-4

t_fixed, y_fixed, fixed_iters = euler_backward_fixed(
    N0, t0, t_final, fixed_h
)

t_adapt, y_adapt, h_hist, error_history, stats = adaptive_implicit_euler(
    y0=N0,
    t0=t0,
    t_final=t_final,
    h0=0.1,
    tol=tol,
    h_min=1e-6,
    h_max=1.0,
    safety=0.9
)

print("Section 4: Performance Analysis\n")
print(f"{'Method':<20} {'Steps':<10} {'NR Iterations':<15}")
print("-" * 50)
print(f"{'Fixed h=0.01':<20} {len(t_fixed)-1:<10} {fixed_iters:<15}")
print(f"{'Adaptive':<20} {stats['accepted_steps']:<10} {stats['total_nr_iterations']:<15}")
print()

print("Adaptive Solver Statistics")
print(f"Accepted steps: {stats['accepted_steps']}")
print(f"Rejected steps: {stats['rejected_steps']}")
print(f"Total steps attempted: {stats['total_steps_attempted']}")
print(f"Acceptance rate: {stats['acceptance_rate']:.2%}")
```

### Comprehensive Comparison of Fixed vs. Adaptive

Euler's Backward Method, with a fixed step size throughout the simulation, is simple and predictable but inefficient for stiff systems. Because the solution contains both fast and slow dynamics, the chosen step size must be small enough to resolve the fastest behavior, even when the solution is changing slowly. This results in many unnecessary steps and increased computational cost. In contrast, the adaptive method adjusts the step size based on an error estimate from step-doubling. It takes smaller steps during the initial transient, where the solution changes very quickly, and then takes larger steps once the solution begins to follow the slowly varying carrying capacity. This leads to a more efficient distribution of time points and reduces the total number of steps required.

In terms of performance and accuracy, the adaptive method is more efficient while maintaining comparable solution quality. Although each adaptive step is more computationally expensive due to multiple Newton-Raphson solves, the reduction in total steps often results in fewer overall computations. Additionally, the adaptive method directly controls error through a tolerance parameter, making it easier to achieve a desired level of accuracy. Therefore, while the fixed step method is simple, the adaptive approach is better suited for stiff systems because it better maintains both accuracy and efficiency.

## Section 6: Conclusions
Damping is necessary when the Newton-Raphson method encounters poor initial guesses, large step sizes, or regions where the derivative is small and the standard Newton step becomes too large. In these cases, damping prevents overshooting and improves convergence by taking more calculated steps.

Adaptive step sizes are worth the added complexity when working with stiff systems, where the solution changes rapidly in some regions and slowly in others. By adjusting the step size based on an error estimate, the method maintains accuracy while avoiding unnecessary computations.

Overall, I learned that balancing accuracy and efficiency often requires more advanced computational techniques, and that methods like damping and adaptive stepping allow solvers to handle dynamics like stiff systems reliably without sacrificing correctness.
