```python
# Kai Gandingco (kcg2279)
# COE 311K
# 9 February 2026

import numpy as np
import matplotlib.pyplot as plt

# Function for calculating pi w/ the Monte Carlo method
def monte_carlo(numSamples):
    x = np.random.uniform(-1,1,size=numSamples)
    y = np.random.uniform(-1,1,size=numSamples)
    counter = ((x*x + y*y) < 1)
    return 4 * (np.sum(counter)/numSamples)
```

```python
error = []

# Iterates number of samples from 10 to 1000 (step size 20)
for x in range (10,1000,20):
    computedPi = monte_carlo(x)
    error.append(np.abs(np.pi - computedPi))
    
# Plots error
samples = np.arange(10,1000,20)
plt.plot(samples, error, color="hotpink")
plt.xlabel("Number of Samples")
plt.ylabel("Absolute Error")
plt.title("Absolute Error vs. Number of Samples")
plt.show()
```
