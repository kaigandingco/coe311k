```python
def myLeastSquares(x, y):
    sum_x = np.sum(x)
    sum_y = np.sum(y)
    sum_x2 = np.sum(x*x)
    sum_xy = np.sum(x*y)
    mNumerator = len(x) * sum_xy - (sum_x * sum_y)
    mDenominator = len(x) * sum_x2 - (sum_x * sum_x)
    m = mNumerator / mDenominator
    c = (sum_y - m * sum_x) / len(x)
    return m, c
```

```python
# Randomized Vectors
X = np.sort(np.random.randint(0, 10, size=10))
Y = np.random.randint(0, 10, size=10)
A = np.vstack([X, np.ones(len(X))]).T

my_m, my_c = myLeastSquares(X, Y)
np_m, np_c = np.linalg.lstsq(A, Y, rcond=None)[0]

# Comparing my Least Squares method with NumPy's
print("Comparing Slopes")
print(f"My Method's Slope: {my_m}\nNumPy's Slope: {np_m}\n")
print("Comparing Intercepts")
print(f"My Method's Intercept: {my_c}\nNumPy's Intercept: {np_c}\n")

# Printing the residuals and identifying points
residuals = Y - (my_m * X + my_c)
print("Residuals from my Least Squares Method:")
print(residuals)
maxResIndex = np.argmax(np.abs(residuals))
minResIndex = np.argmin(np.abs(residuals))
print("\nThe point with the largest residual error was " + 
     f"Point {maxResIndex} at ({X[maxResIndex]}, {Y[maxResIndex]}).")
print("The point with the smallest residual error was " + 
     f"Point {minResIndex} at ({X[minResIndex]}, {Y[minResIndex]}).")

# Plotting
plt.figure()
plt.scatter(X, Y)
plt.plot(X, my_m * X + my_c, label="My Method's Best-Fit Line")
plt.plot(X, np_m * X + np_c, label="NumPy's Best-Fit Line")
plt.title("Least Squares Method with Random Points")
plt.legend()
plt.grid()
plt.show()
```

```python
# Random Linear Equation (y = 5.3x + 2.4)
X = np.arange(0, 1, 0.01)
randPerturb = np.random.normal(0, 1, size=len(X))
Y = 5.3 * X + 2.4 + randPerturb
A = np.vstack([X, np.ones(len(X))]).T

my_m, my_c = myLeastSquares(X, Y)
np_m, np_c = np.linalg.lstsq(A, Y, rcond=None)[0]

# Comparing my Least Squares method with NumPy's
print("Comparing Slopes")
print(f"My Method's Slope: {my_m}\nNumPy's Slope: {np_m}\n")
print("Comparing Intercepts")
print(f"My Method's Intercept: {my_c}\nNumPy's Intercept: {np_c}\n")

# Printing the residuals and identifying points
residuals = Y - (my_m * X + my_c)
print("Residuals from my Least Squares Method:")
print(residuals)
maxResIndex = np.argmax(np.abs(residuals))
minResIndex = np.argmin(np.abs(residuals))
print("\nThe point with the largest residual error was " + 
     f"Point {maxResIndex} at ({X[maxResIndex]}, {Y[maxResIndex]}).")
print("The point with the smallest residual error was " + 
     f"Point {minResIndex} at ({X[minResIndex]}, {Y[minResIndex]}).")

# Plotting
plt.figure()
plt.scatter(X, Y)
plt.plot(X, my_m * X + my_c, label="My Method's Best-Fit Line")
plt.plot(X, np_m * X + np_c, label="NumPy's Best-Fit Line")
plt.title("Least Squares Method with Random Perturbation")
plt.legend()
plt.grid()
plt.show()
```
