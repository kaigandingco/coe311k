```python
def sineTaylor(x, numTerms):
    sumTerms = 0
    for n in range(numTerms):
        exponent = (2 * n) + 1
        sumTerms += (((-1) ** n) * (x ** exponent)) / math.factorial(exponent)
    return sumTerms
```

```python
x = np.arange(0.0, 10.0, 0.1)
for n in range (1,5):
    sineApprox = sineTaylor(x, n)
    error = np.abs(sineApprox - np.sin(x))
    plt.plot(x, error, label=str(n) + " Terms")

# Additional plot formatting
plt.xlabel("x")
plt.ylabel("Absolute Error (scaled)")
plt.yscale("log")
plt.legend()
plt.grid()
plt.title("Truncation Error of the Taylor Series Approximation of sin(x) about 0")
plt.show()
```
